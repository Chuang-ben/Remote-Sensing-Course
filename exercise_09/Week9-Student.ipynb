````python
# [S2] Search and Load Three-Act Scenes
# ──────────────────────────────────────────────────────────────────
# Search STAC for Pre, Mid, Post scenes of Matai'an barrier lake

# Scene dates (three acts of the disaster narrative)
SCENE_DATES = {
    'Pre': '2025-06-01',    # Baseline (before typhoon)
    'Mid': '2025-08-15',    # Lake present at peak
    'Post': '2025-10-10'    # Lake drained / recovered
}

# TODO: Implement robust STAC search
# Hints:
#   1. Create STAC client (e.g., Sentinel-2 L2A collections)
#   2. Search for each date within MATAIAN_BBOX
#   3. Use robust_search() helper to handle failed queries
#   4. Store results in dict: scenes = {'Pre': {...}, 'Mid': {...}, 'Post': {...}}

# Example stub (replace with working code):
def robust_search(client, bbox, start_date, end_date, collection='sentinel-2-l2a'):
    try:
        search = client.search(
            collections=[collection],
            bbox=bbox,
            datetime=f'{start_date}T00:00:00Z/{end_date}T23:59:59Z',
            max_items=10   # 建議多抓幾張
        )

        items = list(search.items())  # ✅ 直接拿 Item（關鍵）
        return items

    except Exception as e:
        print(f"Search failed: {e}")
        return None

# TODO: Complete the search logic
scenes = {}

# Create STAC client
try:
    # Use Microsoft Planetary Computer STAC API
    stac_client = Client.open("https://planetarycomputer.microsoft.com/api/stac/v1")
    print("✓ STAC client connected")
except Exception as e:
    print(f"Failed to connect to STAC client: {e}")
    stac_client = None

# Search for each scene
if stac_client:
    for act, date in SCENE_DATES.items():
        print(f"Searching {act} ({date})...")
        # Create date range (±3 days from target date)
        from datetime import datetime, timedelta
        target_date = datetime.strptime(date, '%Y-%m-%d')
        start_date = (target_date - timedelta(days=15)).strftime('%Y-%m-%d')
        end_date = (target_date + timedelta(days=15)).strftime('%Y-%m-%d')
        
        # Perform robust search
        result = robust_search(stac_client, MATAIAN_BBOX, start_date, end_date)
        
        if result and len(result) > 0:
            # ✅ 選雲量最低（重要）
            best = sorted(
                result,
                key=lambda x: x.properties.get('eo:cloud_cover', 100)
            )[0]

            scenes[act] = best  # ✅ 直接存 Item
            scene_id = best.id
            cloud = best.properties.get('eo:cloud_cover', 'N/A')

            print(f"  ✓ Found scene: {scene_id} (cloud: {cloud}%)")
        else:
            print(f"  ✗ No scenes found for {act}")
else:
    print("✗ STAC client not available - skipping search")

print("✓ Scene search complete")
print(f"  Scenes found: {list(scenes.keys())}")
````

````python
# [S3] Load and Composite Multi-Spectral Cubes
# ──────────────────────────────────────────────────────────────────
# TODO: Implement stream_cube() to load Sentinel-2 bands and create stretched composite

# Reference bands: B02 (Blue), B03 (Green), B04 (Red), B08 (NIR), B11 (SWIR)
BANDS = ['B02', 'B03', 'B04', 'B08', 'B11']
BAND_NAMES = {
    'B02': 'Blue',
    'B03': 'Green',
    'B04': 'Red',
    'B08': 'NIR',
    'B11': 'SWIR'
}

# TODO: Define stream_cube() function
def stream_cube(scene_item, bands, bbox=MATAIAN_BBOX):
    '''Load bands from scene item and stack into cube (H, W, C)'''
    import planetary_computer as pc
    
    # Sign the item for access
    signed_item = pc.sign(scene_item)
    
    # Load all bands at once using odc-stac
    data = stac_load(
        [signed_item],
        bands=bands,
        bbox=bbox,
        epsg=32651,  # UTM zone 51N for Taiwan
        resolution=10,
        dtype="float32"
    )
    
    # Convert to numpy array and remove time dimension
    cube = data.isel(time=0).to_array().values.transpose(1, 2, 0)
    
    # Scale to [0, 1] range (Sentinel-2 DN values are typically scaled by 10000)
    return cube / 10000.0

# TODO: Define composite_stretched() function
def composite_stretched(cube, r_idx=2, g_idx=1, b_idx=0, q=(2, 98)):
    '''Create RGB composite with percentile-based contrast stretch'''
    rgb = cube[:, :, [r_idx, g_idx, b_idx]].copy()
    for i in range(3):
        p_lo, p_hi = np.percentile(rgb[:,:,i], q)
        rgb[:,:,i] = np.clip((rgb[:,:,i] - p_lo) / (p_hi - p_lo), 0, 1)
    return rgb

# TODO: Load and composite each scene
cubes = {}
composites = {}

# Only proceed if we have scenes from S2
if scenes:
    for act in ['Pre', 'Mid', 'Post']:
        if act in scenes:
            print(f"Loading {act}...")
            try:
                cubes[act] = stream_cube(scenes[act], BANDS)
                composites[act] = composite_stretched(cubes[act])
                print(f"  ✓ {act} cube shape: {cubes[act].shape}")
            except Exception as e:
                print(f"  ✗ Failed to load {act}: {e}")
        else:
            print(f"  ✗ No scene data for {act}")
else:
    print("  ✗ No scenes available - run S2 first")

# ── Cloud masking (CRITICAL — without this, ΔNDWI shows "phantom water") ──
# TODO: Stream SCL for each scene and create masks
def stream_scl(scene_item, bbox=MATAIAN_BBOX):
    """Stream SCL and return boolean clear-sky mask."""
    import planetary_computer as pc
    
    signed_item = pc.sign(scene_item)
    scl = stackstac.stack(
        [signed_item], assets=["SCL"], epsg=32651,
        resolution=10, bounds_latlon=bbox, chunksize=2048
    ).squeeze("time").squeeze("band")
    
    # Compute and return clear-sky mask
    scl_data = scl.compute().values
    return np.isin(scl_data, SCL_CLEAR_CLASSES)

def safe_compute(data):
    """Safely compute dask array."""
    return data.compute()

# Apply cloud masking if cubes are available
clear_masks = {}
if cubes:
    for act in ['Pre', 'Mid', 'Post']:
        if act in scenes:
            print(f"Creating cloud mask for {act}...")
            try:
                clear_masks[act] = stream_scl(scenes[act])
                clear_ratio = np.sum(clear_masks[act]) / clear_masks[act].size
                print(f"  ✓ Clear sky ratio: {clear_ratio:.2%}")
            except Exception as e:
                print(f"  ✗ Failed to create mask for {act}: {e}")
                clear_masks[act] = np.ones_like(list(cubes.values())[0][:, :, 0], dtype=bool)

# ── Per-scene valid masks (for Three-Act visualization) ──
# Each scene uses its OWN mask so students can see Mid has more clouds.
valid_masks = {}
if cubes and clear_masks:
    for act in ['Pre', 'Mid', 'Post']:
        if act in cubes and act in clear_masks:
            # Valid if finite values AND clear sky
            valid_masks[act] = np.isfinite(cubes[act]).all(axis=2) & clear_masks[act]
            valid_ratio = np.sum(valid_masks[act]) / valid_masks[act].size
            print(f"  {act} valid pixel ratio: {valid_ratio:.2%}")

# ── Intersection mask (for difference maps where BOTH dates must be valid) ──
valid_intersection = None
if valid_masks and len(valid_masks) >= 2:
    # Intersection of all available valid masks
    valid_intersection = np.ones_like(list(valid_masks.values())[0], dtype=bool)
    for mask in valid_masks.values():
        valid_intersection &= mask
    intersection_ratio = np.sum(valid_intersection) / valid_intersection.size
    print(f"  Intersection valid ratio: {intersection_ratio:.2%}")

print("✓ Cubes loaded and composited")
print(f"  Cube shape: (H, W, C) = {list(cubes.values())[0].shape if cubes else 'N/A'}")
````

### Computing Spectral Indices: NDVI & NDWI
### 計算光譜指標：NDVI & NDWI

**NDVI** = (NIR - Red) / (NIR + Red)
**NDWI** = (Green - NIR) / (Green + NIR)

─

Why two indices?
- **NDVI**: Sensitive to vegetation density, but also responds to water
- **NDWI**: Maximally sensitive to open water and wetlands

─

為何使用兩種指標？
- **NDVI**：對植被濃度敏感，但也對水體有反應
- **NDWI**：對開放水體和濕地最敏感


````python
# [S4] Compute NDVI and NDWI for All Three Scenes
# ──────────────────────────────────────────────────────────────────

def ndvi(cube):
    red = cube[:, :, 2]
    nir = cube[:, :, 3]
    return (nir - red) / (nir + red + 1e-8)

def ndwi(cube):
    green = cube[:, :, 1]
    nir = cube[:, :, 3]
    return (green - nir) / (green + nir + 1e-8)

indices = {}

if cubes:
    for act in ['Pre', 'Mid', 'Post']:
        if act in cubes:
            print(f"Computing indices for {act}...")
            try:
                ndvi_arr = ndvi(cubes[act]).astype(np.float32)
                ndwi_arr = ndwi(cubes[act]).astype(np.float32)

                # Apply valid_mask: set invalid (cloud/no-data) pixels to NaN
                if act in valid_masks:
                    mask = valid_masks[act]
                    ndvi_arr[~mask] = np.nan
                    ndwi_arr[~mask] = np.nan

                indices[act] = {'NDVI': ndvi_arr, 'NDWI': ndwi_arr}
                print(f"  ✓ {act} indices computed")
            except Exception as e:
                print(f"  ✗ Failed to compute {act} indices: {e}")
        else:
            print(f"  ✗ No cube data for {act}")
else:
    print("  ✗ No cubes available - run S3 first")

if indices:
    for act in ['Pre', 'Mid', 'Post']:
        if act in indices:
            print(f"{act} scene:")
            for idx_name in ['NDVI', 'NDWI']:
                idx = indices[act][idx_name]
                print(f"  {idx_name}: μ={np.nanmean(idx):.3f}, σ={np.nanstd(idx):.3f}, "
                      f"[{np.nanmin(idx):.3f}, {np.nanmax(idx):.3f}]")
        else:
            print(f"{act} scene: No data available")

print("✓ NDVI and NDWI computed with valid_mask applied")
````

### Discussion: Threshold is a DECISION, Not a Formula
### 討論：臨界值是「決策」，不是「公式」

**Key insight**: No single "correct" threshold exists. The choice depends on your **use case**:

| Use Case | Threshold | Rationale | 優先考量 |
|----------|-----------|-----------|---------|
| **Disaster Alert** | Low (0.05) | Catch early signals, high false positive OK | 敏感度 |
| **Insurance Payout** | Medium (0.10) | Balance detection & accuracy | 平衡 |
| **Scientific Archive** | High (0.15) | Minimize false positives | 特異性 |

─

**Question for reflection**:
- In the Oct 2025 post-event phase, should ARIA v6.0 use a low or high threshold?
    - In the post-event phase, ARIA v6.0 should use a higher threshold (e.g., 0.10–0.15) to ensure reliable detection of remaining water. This reduces false positives caused by wet soil, shadows, or noise, providing more accurate information for recovery assessment.
- What are the consequences of each choice for emergency responders?
    - Using a low threshold increases sensitivity but leads to many false positives, which may cause emergency responders to overestimate flood extent and misallocate resources. In contrast, a high threshold improves reliability by reducing false alarms, allowing for more accurate decision-making during recovery, although it may miss some marginal water areas.

````python
# [S9] Compute Confusion Matrix
# ──────────────────────────────────────────────────────────────────
# TODO: Sample your detection mask at ground truth locations.
#    For now, a demo confusion matrix is shown below.
# ──────────────────────────────────────────────────────────────────

# TODO: In real workflow, sample predicted mask at validation points:
# validation_points['predicted'] = [mask[geo_to_pixel(lon, lat, gt)] ...]
# cm = confusion_matrix(ground_truth, predicted, labels=[0, 1])

# ── Synthetic confusion matrix for demo ──
predicted = np.random.randint(0, 2, len(validation_points))
ground_truth = validation_points['truth'].values

cm = np.array([[12, 3],   # TN=12, FP=3
               [2, 13]])   # FN=2, TP=13

print("✓ Confusion matrix computed (demo values)")
print("\nConfusion Matrix:")
print(cm)
print("\nLabels: [0=No Water, 1=Water]")
print(f"  TN (True Negatives): {cm[0,0]}")
print(f"  FP (False Positives): {cm[0,1]}")
print(f"  FN (False Negatives): {cm[1,0]}")
print(f"  TP (True Positives): {cm[1,1]}")
````

````python
# [S10] Compute Accuracy Metrics
# ──────────────────────────────────────────────────────────────────
# TODO: Calculate Producer's Accuracy, User's Accuracy, Overall Accuracy, Kappa

# Extract confusion matrix elements
tn, fp, fn, tp = cm[0,0], cm[0,1], cm[1,0], cm[1,1]

# TODO: Producer's Accuracy (Sensitivity, Recall, True Positive Rate)
producer_accuracy = tp / (tp + fn)

# TODO: User's Accuracy (Precision, Positive Predictive Value)
user_accuracy = tp / (tp + fp)

# TODO: Overall Accuracy
overall_accuracy = (tp + tn) / (tp + tn + fp + fn)

# TODO: Cohen's Kappa (accounts for chance agreement)
# po = overall_accuracy (observed agreement)
# pe = expected agreement by chance
pe = ((tp + fn) * (tp + fp) + (tn + fp) * (tn + fn)) / (tp + tn + fp + fn)**2
kappa = (overall_accuracy - pe) / (1 - pe)

print("✓ Accuracy metrics computed\n")
print("=" * 50)
print("ARIA v6.0 VALIDATION METRICS")
print("=" * 50)
print(f"Producer's Accuracy (Sensitivity): {producer_accuracy:.3f} ({producer_accuracy*100:.1f}%)")
print(f"User's Accuracy (Precision):       {user_accuracy:.3f} ({user_accuracy*100:.1f}%)")
print(f"Overall Accuracy:                  {overall_accuracy:.3f} ({overall_accuracy*100:.1f}%)")
print(f"Cohen's Kappa:                     {kappa:.3f}")
print("=" * 50)

# Store metrics for later use
metrics = {
    'producer_accuracy': producer_accuracy,
    'user_accuracy': user_accuracy,
    'overall_accuracy': overall_accuracy,
    'kappa': kappa,
    'tp': tp, 'tn': tn, 'fp': fp, 'fn': fn
}
````

### Discussion: What Do These Accuracy Numbers Mean for Disaster Response?
### 討論：這些精度數字對災難應對意味著什麼？

Consider ARIA's role in Typhoon Colo response (Aug 2025):

1. **Producer's Accuracy (Sensitivity) = 87%**
   - Out of all actual water pixels, we correctly identified 87%
   - Missing 13% of the true lake area
   - Implication: Evacuation zone might miss some flooded areas

2. **User's Accuracy (Precision) = 81%**
   - Out of all pixels we called "water", 81% actually are water
   - 19% false alarms (crying wolf)
   - Implication: Emergency responders waste resources on false alarms

3. **Overall Accuracy = 83%**
   - 83% of all predictions correct
   - Looks good, but hides class imbalance if water is rare

─

**Critical question**: In a disaster setting, which error is worse?
- **False Negative** (missing flooded area) → people in danger
- **False Positive** (false alarm) → wasted resources, lost credibility

For early warning systems, we often prefer **high sensitivity** (low false negatives) over precision.


### Discussion: Why Is Producer's Accuracy More Important Than Overall Accuracy?
### 討論：為何生產者精度比整體精度更重要？

**Scenario**: Matai'an barrier lake covers ~10% of the study area. Rest is land/forest.

If model predicts: "Everything is NOT water"
- TN = 90%, FP = 0%, TP = 0%, FN = 10%
- **Overall Accuracy** = 90% ✓ (looks amazing!)
- **Producer's Accuracy (Sensitivity)** = 0% ✗ (catastrophic: missed all the water!)

This is called the **accuracy paradox** or **class imbalance problem**.

─

**For disaster detection**:
- We care most about **not missing disasters** → Prioritize **Sensitivity / Producer's Accuracy**
- Trade-off: Accept some false positives (precision tradeoff)
- This is why early warning systems often have lower precision but higher sensitivity

**ARIA v6.0 design principle**:
> "Better to alert 10 safe neighborhoods than miss 1 at-risk community."
